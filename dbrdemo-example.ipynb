## Use Spark and dbutils

Import the connected Spark session and dbutils from `dbrdemo`. The example checks the current Databricks identity and catalog, then lists secret scopes.

```python
from dbrdemo import dbutils, spark

# use spark
spark.sql(
    "SELECT current_user() AS user, current_catalog() AS catalog"
).show(truncate=False)

# list secret scopes
dbutils.secrets.listScopes()
```

## Create and write foo/bar data

Create a DataFrame first. Before writing it, change the table name to a Unity Catalog schema where you have write access.

```python
from dbrdemo import create_foobar, write_foobar

foobar = create_foobar("from", "notebook")
foobar.show(truncate=False)

# change table name to a Unity Catalog schema where you have write access
table_name = "main.demo.foobar"
write_foobar(table_name, "from", "notebook")
spark.table(table_name).show(truncate=False)
```
