## Use Spark and dbutils

Import the connected Spark session and dbutils from `dbrdemo`. The example checks the current identity and catalog, then lists secret scopes.

```python
from dbrdemo import dbutils, install_logger, spark

install_logger()  # Optional: show events emitted by the library.

# use spark
spark.sql(
    "SELECT current_user() AS user, current_catalog() AS catalog"
).show(truncate=False)

# list secret scopes
dbutils.secrets.listScopes()
```
